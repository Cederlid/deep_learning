**Set initial (random) values for weights and biases of our neuron**

```python
# set initial (random) values for weights and biases of our neuron

w_1 = 0.3
w_2 = 0.5
b = 2
```

**Define our features and targets $y_{true}$ for the training samples in our batch**

```python
x_1_1 = -2     # första feature för sample 1
x_2_1 = 1.5    # andra feature för sample 1

x_1_2 = -5     # första feature för sample 2
x_2_2 = 2      # andra feature för sample 2

y_true_1 = 10  # y_true för sample 1
y_true_2 = 6   # y_true för sample 2
```

**Set our learning rate**

```python
learning_rate = 0.01
```

**Do all needed calculations (loss and gradients), and perform gradient descent**

```python
y_pred_1 = x_1_1*w_1 + x_2_1*w_2 + b
y_pred_2 = x_1_2*w_1 + x_2_2*w_2 + b

print('Prediction for sample 1 :', y_pred_1)
print('Prediction for sample 2 :', y_pred_2)

loss_mse = 0.5*( (y_true_1 - y_pred_1)**2 + (y_true_2 - y_pred_2)**2)

print('MSE Loss                :', loss_mse)

loss_gradient_w_1 = -x_1_1*(y_true_1 - y_pred_1) - x_1_2*(y_true_2 - y_pred_2)
loss_gradient_w_2 = -x_2_1*(y_true_1 - y_pred_1) - x_2_2*(y_true_2 - y_pred_2)
loss_bias = -(y_true_1 - y_pred_1) - (y_true_2 - y_pred_2)

print(f'loss_gradient_w_1       : {loss_gradient_w_1}')
print(f'loss_gradient_w_2       : {loss_gradient_w_2}')
print(f'loss_bias               : {loss_bias}')

w_1 = w_1 - learning_rate*loss_gradient_w_1
w_2 = w_2 - learning_rate*loss_gradient_w_2
b = b - learning_rate*loss_bias

print(f'w_1_new                 : {w_1}')
print(f'w_2_new                 : {w_2}')
print(f'b_new                   : {b}')
```
